```python
import pandas as pd
import numpy as np
from scipy.stats import trim_mean
from statsmodels import robust
import wquantiles
```

```python
STATE_CSV = '/Users/rukku/Documents/Sheridan/Python/PythonProject/stateDataset.csv'
state = pd.read_csv(STATE_CSV)
print(state.head(8))
```

```python
print(print("Population Mean:", state['Population'].mean()))
print(trim_mean(state['Population'], 0.1))
print(state['Population'].median())
print(state['Murder.Rate'].mean())
print(np.average(state['Murder.Rate'], weights=state['Population']))
print(wquantiles.median(state['Murder.Rate'], weights=state['Population']))
```

```python
print(state['Population'].std())
print(state['Population'].quantile(0.75) - state['Population'].quantile(0.25))
print(robust.scale.mad(state['Population']))
print(abs(state['Population'] - state['Population'].median()).median() / 0.6744897501960817)
```

```python
iris_csv = '/Users/rukku/Documents/Sheridan/Python/PythonProject/iris.csv'
iris = pd.read_csv(iris_csv)
print(iris.head(8))
```

```python
print(iris['sepal_length'].median())
print(iris['sepal_length'].mean())
print(iris['sepal_length'].std())
print(robust.scale.mad(iris['sepal_length']))
print(abs(iris['sepal_length'] - iris['sepal_length'].median()).median() / 0.6744897501960817)
q75, q25 = np.percentile(iris["sepal_length"], [75, 25], method='linear')  # matches R type 7
iqr = q75 - q25
print("Interquartile Range (IQR):", iqr)
```
