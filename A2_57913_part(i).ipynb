```python
import random
import numpy as np
import matplotlib.pyplot as plt
from math import comb

# Hardcoded Binomial PMF
def binomial_pmf(k, n, p):
    return comb(n, k) * (p**k) * ((1-p)**(n-k))

# Monte Carlo Simulation
def monte_carlo(num_trials):

    results = []

    for i in range(num_trials):

        success = 0

        for j in range(25):

            r = random.random()

            if r < 0.84:
                success += 1

        results.append(success)

    return results

# Run 1000 simulations
mc_1000 = monte_carlo(1000)

# Theoretical Distribution
x = np.arange(0, 26)

y = []

for k in x:
    y.append(binomial_pmf(k, 25, 0.84))

# Figure 3
plt.figure(figsize=(10,6))

plt.hist(
    mc_1000,
    bins=50,
    density=True,
    alpha=0.6,
    color='skyblue',
    label='Monte Carlo (1000 iterations)'
)

plt.plot(
    x,
    y,
    'ro-',
    linewidth=2,
    label='Theoretical Distribution'
)

plt.title('Fig.3 Monte Carlo vs Theoretical Distribution')
plt.xlabel('Number of Correct Detections')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)

plt.show()
```
