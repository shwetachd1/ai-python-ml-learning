```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
scores = np.array([
    88, 78, 64, 92, 57, 70, 88, 68, 72, 60, 60, 73, 85, 89, 73, 52, 71, 51,
    73, 93, 79, 87, 51, 70, 82, 61, 71, 93, 74, 98, 76, 91, 77, 65, 64, 96,
    100, 93, 52, 86, 100, 56, 70, 58, 88, 67, 53, 74, 63, 99, 58, 75, 51, 69,
    77, 96, 56, 93, 57, 96, 84, 63, 66, 85, 99, 89, 53, 51, 55, 91, 53, 78,
    67, 75, 93, 83, 59, 85, 63, 80, 97, 64, 57, 63, 72, 89, 70, 65, 94, 67,
    96, 73, 75, 74, 94, 90, 78, 64, 94, 50
])

df = pd.DataFrame(scores, columns=["Test Scores"])
print(df)
```

```python
print(df['Test Scores'].median())
print(df['Test Scores'].mean())
q75, q25 = np.percentile(df["Test Scores"], [75, 25], method='linear')
print(q75, q25)
percentages = [25, 50,75]
for p in percentages:
    percentile_value = np.percentile(df, p)
    print(f"{p}th percentile: {percentile_value}")
```

```python
ax= df.plot.box(title= 'Box Plot of Test Scores', figsize=[6,4])
ax.set_ylabel('Score Value')
plt.tight_layout()
plt.show()
```

```python
binnedScores = pd.cut(df['Test Scores'], 5)
print(binnedScores.value_counts())
```

```python
ax = (df['Test Scores'].plot.hist(figsize=(4,4), bins = 10, color ='lightgreen', edgecolor='black', linewidth=1.5))
ax.set_xlabel('Scores')
plt.tight_layout()
plt.show()
```

```python
ax = df['Test Scores'].plot.hist(density=True, xlim=[0,100], bins=range(0,101), figsize=(4,4), alpha=.05,color='lightgreen')
df['Test Scores'].plot.density(ax=ax)
ax.set_xlabel('Test Scores')
plt.tight_layout()
plt.show()
```
