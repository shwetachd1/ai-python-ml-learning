```python
from gettext import install

import pandas as pd
import numpy as np
from scipy.stats import trim_mean
from statsmodels import robust
import wquantiles
```

```python
df = pd.DataFrame({
    "State": ["Alabama", "Alaska", "Arizona", "Arkansas", "California",
              "Colorado", "Connecticut", "Delaware"],
    "Population": [4779736, 710231, 6392017, 2915918, 37253956,
                   5029196, 3574097, 897934],
    "MurderRate": [5.7, 5.6, 4.7, 5.6, 4.4, 2.8, 2.4, 5.8]
})

print (df)

```

```python
print("Population Mean:", df['Population'].mean())
print("Median:", df['Population'].median())

df2 = df.sort_values(by='Population')
print(df2)
trim_count = int((len(df2) *0.1))
trim_df2 = df.iloc[trim_count : -trim_count]
print(trim_df2)

print(df['Population'])
```

```python
# Mean 
print("Population Mean-Population:", df['Population'].mean())

# Weighted Mean: Murder Rate with Population weight
print("Weighted Mean-Murder Rate and Population weights:", np.average(df['MurderRate'], weights=df['Population']))

# Trim-Mean: drops 10% from each end
print("Trim-mean- Population", trim_mean(df['Population'], 0.1))

# Mean, Median : Murder Rate
print("Mean  - Murder Rate   : ", df["MurderRate"].mean())
print("Median- Murder Rate :", df["MurderRate"].median())
```

```python
print(np.average(df['MurderRate'], weights=df['Population']))

print(wquantiles.median(df['MurderRate'], weights=df['Population']))

print(df['Population'].std())
print(df['Population'].quantile(0.75) - df['Population'].quantile(0.25))
print(robust.scale.mad(df['Population']))
print(abs(df['Population'] - df['Population'].median()).median() / 0.6744897501960817)
```

```python
# Standard Deviation
std_dev = df["Population"].std(ddof=1)
# Interquartile Range (IQR)
q75, q25 = np.percentile(df["Population"], [75, 25], method='linear')  # matches R type 7
iqr = q75 - q25

# Median Absolute Deviation (MAD), scaled like R
median_population = df["Population"].median()
mad = np.median(np.abs(df["Population"] - median_population)) * 1.4826

# Print all results
print("Standard Deviation:", std_dev)
print("Interquartile Range (IQR):", iqr)
print("Median Absolute Deviation (MAD):", mad)
```
