```python
Here’s a Python example using house features, a very common real-world dataset:
Dataset Variables
SquareFootage → house size
NumBedrooms → number of bedrooms
Age → age of the house
Price → house price
This reflects real-world correlations:
Bigger house → higher price (positive)
Newer house → higher price (negative correlation with age)
```

```python
np.random.seed(42)

# Create realistic housing dataset
size = 200

square_feet = np.random.normal(1800, 400, size)
bedrooms = np.random.randint(2, 6, size)
age = np.random.randint(0, 30, size)

# Price influenced by size, bedrooms, and age
price = (
    square_feet * 150 +
    bedrooms * 10000 -
    age * 1200 +
    np.random.normal(0, 20000, size)  # noise
)

df = pd.DataFrame({
    "SquareFeet": square_feet,
    "Bedrooms": bedrooms,
    "Age": age,
    "Price": price
})
print(df)
print(df.columns)
```

```python
# Compute correlation matrix
corr_matrix = df.corr()
print (corr_matrix)
```

```python
fig, ax = plt.subplots(figsize=(5, 4))
ax = sns.heatmap(df.corr(), vmin=-1, vmax=1, 
                 cmap=sns.diverging_palette(20, 220, as_cmap=True),
                 ax=ax)
```

```python
ax = df.plot.scatter(x='SquareFeet', y='Price', figsize=(4, 4), marker='$\u25EF$')
ax.set_xlabel('Square Feet')
ax.set_ylabel('Price')
ax.axhline(0, color='grey', lw=1)
ax.axvline(0, color='grey', lw=1)

plt.tight_layout()
plt.show()
```

```python
ax = df.plot.scatter(x='Age', y='Price', figsize=(4, 4), marker='$\u25EF$')
ax.set_xlabel('Square Feet')
ax.set_ylabel('Price')
ax.axhline(0, color='grey', lw=1)
ax.axvline(0, color='grey', lw=1)

plt.tight_layout()
plt.show()
```

```python
ax = df.plot.scatter(x='SquareFeet', y='Age', figsize=(4, 4), marker='$\u25EF$')
ax.set_xlabel('Square Feet')
ax.set_ylabel('Price')
ax.axhline(0, color='grey', lw=1)
ax.axvline(0, color='grey', lw=1)

plt.tight_layout()
plt.show()
```

```python
ax = df.plot.hexbin(x='SquareFeet', y='Price',
                         gridsize=30, sharex=False, figsize=(5, 4))
ax.set_xlabel('Square Feet')
ax.set_ylabel('Price')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4, 4))
sns.kdeplot(data=df.sample(100), x='SquareFeet', y='Price', ax=ax)
ax.set_xlabel('Square Feet')
ax.set_ylabel('Price')

plt.tight_layout()
plt.show()
```

```python
data = {
    'Car Model': ['Focus', 'Mustang', 'F-150', 'Explorer', 'Fusion', 'Fiesta', 'Bronco', 'GT', 'Ranger', 'Escape'],
    'Engine Size (L)': [2.0, 5.0, 3.5, 2.3, 2.5, 1.6, 2.7, 3.0, 2.3, 1.5],
    'Fuel Economy (MPG)': [35, 18, 22, 25, 30, 42, 20, 15, 24, 32],
    'Car Type': ['Sedan', 'Coupe', 'Truck', 'SUV', 'Sedan', 'Sedan', 'SUV', 'Coupe', 'Truck', 'SUV'],
    'Color': ['Red', 'Black', 'Blue', 'White', 'Silver', 'Red', 'Green', 'Black', 'Gray', 'Blue'],
    'Price ($k)': [25, 50, 45, 40, 30, 18, 48, 150, 35, 28]
}

df = pd.DataFrame(data)
```

```python
import pandas as pd
crosstab = pd.crosstab(df['Car Model'], df['Color'], margins=True)
crosstab.style.background_gradient(cmap='Blues')
#print(crosstab)
```

```python
ax = df.boxplot(by='Car Model', column='Price ($k)',figsize=(7, 7))
ax.set_xlabel('Car Model')
ax.set_ylabel('Price')
plt.suptitle('Car Model Vs Price')

plt.tight_layout()
plt.show()
```

```python
colors = ['Red', 'Black', 'Blue', 'White', 'Silver', 'Green', 'Gray']
df_color = df.loc[df.Color.isin(colors),:]

def hexbin(x, y, color, **kwargs):
    cmap = sns.light_palette(color, as_cmap=True)
    plt.hexbin(x, y, gridsize=25, cmap=cmap, **kwargs)

g = sns.FacetGrid(df_color, col='Color', col_wrap=2)
g.map(hexbin, 'Engine Size (L)', 'Fuel Economy (MPG)', 
      extent=[0, 5, 0, 50])
g.set_axis_labels('Engine Size', 'Fuel Economy (MPG)')

g.set_titles('Color {col_name:}')
plt.tight_layout()
plt.show()
```
