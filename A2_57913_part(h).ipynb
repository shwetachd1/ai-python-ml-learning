```python
import random

# Monte Carlo Simulation to estimate the number of successes in 25 trials with a success probability of 0.84

def monte_carlo(num_trials):

    results = []

    for i in range(num_trials):

        success = 0

        for j in range(25):

            r = random.random()

            if r < 0.84:
                success += 1

        results.append(success)

    return results

# Run the Monte Carlo simulation for different numbers of trials
mc_500 = monte_carlo(500)
mc_1000 = monte_carlo(1000)
mc_2000 = monte_carlo(2000)

print("First 10 results from 500  trials:", mc_500[:10])
print("First 10 results from 1000 trials:", mc_1000[:10])
print("First 10 results from 2000 trials:", mc_2000[:10])

print("Mean 500  trials:", sum(mc_500)  / len(mc_500))
print("Mean 1000 trials:", sum(mc_1000) / len(mc_1000))
print("Mean 2000 trials:", sum(mc_2000) / len(mc_2000))

print("length of Results:")
print(len(mc_500))
print(len(mc_1000))
print(len(mc_2000))
```
