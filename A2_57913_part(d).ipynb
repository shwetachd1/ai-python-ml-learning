```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

#hardcoded PMF function for binomial distribution
def binomial_pmf(k, n, p):
    return comb(n, k) * (p**k) * ((1-p)**(n-k))

# Blue
n1 = 25
p1 = 0.84
x1 = np.arange(0, n1+1)
y1 = [binomial_pmf(k, n1, p1) for k in x1]

# Green
n2 = 40
p2 = 0.93
x2 = np.arange(0, n2+1)
y2 = [binomial_pmf(k, n2, p2) for k in x2]

# Red
n3 = 90
p3 = 0.99
x3 = np.arange(0, n3+1)
y3 = [binomial_pmf(k, n3, p3) for k in x3]

plt.figure(figsize=(10,6))

#Using stem plot to visualize the PMF for each distribution
plt.stem(x1, y1, linefmt='b-', markerfmt='bo', basefmt='grey', label='Blue: n=25, p=0.84')
plt.stem(x2, y2, linefmt='g-', markerfmt='go', basefmt='grey', label='Green: n=40, p=0.93')
plt.stem(x3, y3, linefmt='r-', markerfmt='ro', basefmt='grey', label='Red: n=90, p=0.99')



plt.xlabel('Number of Correct Detections')
plt.ylabel('Probability')
plt.title('Fig.1 Binomial Probability Distributions')
plt.legend()
plt.grid(True)

plt.show()
```
