### Create a new figure, Fig.2, and only include the “green” plot described above. Find the mean and spread for this distribution and show it on the plot. Show all your calculations and steps

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

# Hardcoded Binomial PMF
def binomial_pmf(k, n, p):
    return comb(n, k) * (p**k) * ((1-p)**(n-k))

# Green Distribution Parameters
n = 40
p = 0.93

# Generate x and y values
x = np.arange(0, n + 1)

y = []
for k in x:
    y.append(binomial_pmf(k, n, p))

# Mean
mean = n * p

# Variance
variance = n * p * (1 - p)

# Standard Deviation (Spread)
sd = variance ** 0.5

print("Mean =", mean)
print("Variance =", variance)
print("Standard Deviation =", sd)

# Create Figure 2
plt.figure(figsize=(10,6))

plt.plot(x, y,
         color='green',
         marker='o',
         label='n=40, p=0.93')

# Mean
plt.axvline(mean,
            color='black',
            linewidth=2,
            label=f'Mean = {mean:.2f}')

# Mean - SD
plt.axvline(mean - sd,
            color='red',
            linestyle='--',
            label='Mean ± SD')

# Mean + SD
plt.axvline(mean + sd,
            color='red',
            linestyle='--')

plt.title('Fig.2 Green Distribution')
plt.xlabel('Number of Correct Detections')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)

plt.show()
```
