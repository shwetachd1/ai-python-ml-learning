## Part C: Data Cleaning

```python
import pandas as pd
import numpy as np

# Ensure reproducibility
np.random.seed(42)

# Randomly select 5 rows to duplicate
rows_to_duplicate = df.sample(n=5)
df_with_duplicates = pd.concat([df, rows_to_duplicate], ignore_index=True)
# Count duplicates
df_with_duplicates.duplicated().sum()

# Display duplicate rows
df_with_duplicates[df_with_duplicates.duplicated(keep=False)]
```

```python
# TODO: Check for duplicate rows
df.duplicated().any()
# TODO: Remove duplicate rows if any exist
df = df.drop_duplicates()
#verify if it worked
df.duplicated().any()
```
