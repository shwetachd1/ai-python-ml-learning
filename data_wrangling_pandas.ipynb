```python

import pandas as pd
```

## Create Sample Dataset

```python

data = {
    "OrderID": [1001,1002,1003,1004,1005,1006,1007,1008,1008,1009,1010],
    "Customer": ["Ali","Sara","John","Aisha","Mike","Emily","Ali","Sara","Sara","John","Aisha"],
    "Category": ["Electronics","Electronics","Furniture","Electronics","Furniture","Clothing","electronics","Clothing","Clothing","Furniture","Clothing"],
    "Product": ["Laptop","Phone","Chair",None,"Table","Jacket","Tablet","Shoes","Shoes","Sofa","T-Shirt"],
    "Price": [1200,800,150,900,None,200,450,120,120,950,35],
    "Quantity": [1,2,4,1,1,2,1,3,3,1,5],
    "OrderDate": ["2025-01-12","2025-01-15","2025-01-16","2025-01-18","2025-01-20","2025-01-21","2025-01-22","2025-01-25","2025-01-25","2025-01-28","2025-01-29"],
    "City": ["Toronto","Montreal","Vancouver","Toronto","Calgary",None,"Toronto","Montreal","Montreal","Vancouver","Toronto"]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print("Dataset created successfully.")
print(df.head())
```

## Load the Dataset

```python

df = pd.read_csv("sales_data.csv")

df.head()
```

## Explore Dataset Information

```python
df.info()
```

```python
df.describe()
```

## Check Missing Values

```python
df.isnull().sum()
```

## Handle Missing Values

```python

df["Product"] = df["Product"].fillna("Unknown")

average_price = df["Price"].mean()

df["Price"] = df["Price"].fillna(average_price)

df["City"] = df["City"].fillna("Unknown")

df
```

## Detect and Remove Duplicates

```python

print("Duplicate Rows:")
print(df.duplicated())

df = df.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(df.shape)
```

## Standardize Text Values

```python

df["Category"] = df["Category"].str.capitalize()

df["Category"].unique()
```

## Create New Column

```python

df["TotalSales"] = df["Price"] * df["Quantity"]

df.head()
```

## Convert Data Types

```python

df["OrderDate"] = pd.to_datetime(df["OrderDate"])

df.dtypes
```

## Filter Data

```python

electronics_df = df[df["Category"] == "Electronics"]

electronics_df
```

```python

high_sales = df[df["Price"] > 500]

high_sales
```

## Sort Data

```python

sorted_df = df.sort_values(by="TotalSales", ascending=False)

sorted_df
```

## Grouping and Aggregation

```python

category_sales = df.groupby("Category")["TotalSales"].sum()

category_sales
```

```python

average_category_price = df.groupby("Category")["Price"].mean()

average_category_price
```

## Export Cleaned Dataset

```python

df.to_csv("cleaned_sales_data.csv", index=False)

print("Cleaned dataset exported successfully.")
```
