```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Number of samples
n_samples = 100

# Generate 3 random features
X1 = np.random.rand(n_samples) * 10  # Feature 1
X2 = np.random.rand(n_samples) * 5   # Feature 2
X3 = np.random.rand(n_samples) * 3   # Feature 3

# True coefficients
w1, w2, w3 = 2.5, -1.2, 0.8

# Linear relationship + some noise
y = w1*X1 + w2*X2 + w3*X3 + np.random.randn(n_samples) * 0.5

# Combine features into one array
features = np.column_stack((X1, X2, X3))

# Print first row
print("Features:", features[0])
print("Target:", y[0])
```

```python
plt.scatter(features[:, 0], y)
plt.xlabel("Feature 1")
plt.ylabel("Target")
plt.title("Feature 1 vs Target")
plt.show()
```

If we are interested in creating a simulated dataset for classification, we can use `make_classification`:

```python
# load library
from sklearn.datasets import make_classification
# generate features matrix and target vector
features, target = make_classification(n_samples = 100,
                                       n_features = 3,
                                       n_informative = 3,
                                       n_redundant = 0,
                                       n_classes = 2,
                                       weights = [.25, .75],
                                       random_state = 1)
# view feature matrix and target vector
print("Feature matrix\n {}".format(features[:3]))
print("Target vector\n {}".format(target[:3]))
```

```python
import matplotlib.pyplot as plt

plt.scatter(features[:,0],target, c=target)
plt.xlabel("Feature 0")
plt.ylabel("Target")
plt.title(f"Feature 0 vs Target")
plt.show()
```

```python
# load library
from sklearn.datasets import make_blobs

# generate feature_matrix and target vector
features, target = make_blobs(n_samples = 100,
                              n_features = 2,
                              centers = 3,
                              cluster_std = 0.5,
                              shuffle = True,
                              random_state = 1)

# view feature matrix and target vector
print("Feature Matrix\n {}".format(features[:3]))
print("Target Vector\n {}".format(target[:3]))
```

```python
# ============================================================
# COMPLETE SQLITE + SQLALCHEMY + PANDAS EXAMPLE
# ============================================================
#
# This example demonstrates:
#
# 1. Creating SQLite database
# 2. Creating tables
# 3. Inserting sample data
# 4. Reading existing data using pandas
# 5. Filtering data
# 6. Reading specific columns
# 7. Displaying table names
# 8. Performing basic data analysis
#
# Required Libraries:
# pip install pandas sqlalchemy
#
# ============================================================

from sqlalchemy import (
    create_engine,
    MetaData,
    Table,
    Column,
    Integer,
    String,
    Float,
    inspect,
    text
)

import pandas as pd


# ============================================================
# STEP 1 — CONNECT TO SQLITE DATABASE
# ============================================================

# Creates company.db file if it does not exist
engine = create_engine("sqlite:///company.db")

print("Connected to SQLite database successfully!")


# ============================================================
# STEP 2 — CREATE TABLES
# ============================================================

metadata = MetaData()

employees = Table(
    "employees",
    metadata,

    Column("id", Integer, primary_key=True),
    Column("name", String),
    Column("department", String),
    Column("salary", Float)
)

# Create tables inside database
metadata.create_all(engine)

print("Tables created successfully!")


# ============================================================
# STEP 3 — INSERT SAMPLE DATA
# ============================================================

sample_data = [
    {"id": 1, "name": "John", "department": "IT", "salary": 70000},
    {"id": 2, "name": "Sara", "department": "HR", "salary": 65000},
    {"id": 3, "name": "Mike", "department": "Finance", "salary": 80000},
    {"id": 4, "name": "Emma", "department": "IT", "salary": 90000},
    {"id": 5, "name": "David", "department": "Marketing", "salary": 60000},
]

# Insert data into table
with engine.connect() as connection:

    # Clear old data first
    connection.execute(text("DELETE FROM employees"))

    # Insert new data
    connection.execute(employees.insert(), sample_data)

    # Commit transaction
    connection.commit()

print("Sample data inserted successfully!")


# ============================================================
# STEP 4 — SHOW EXISTING TABLES
# ============================================================

inspector = inspect(engine)

tables = inspector.get_table_names()

print("\nAvailable Tables:")
print(tables)


# ============================================================
# STEP 5 — READ EXISTING DATA USING PANDAS
# ============================================================

print("\n================================================")
print("READING ENTIRE EMPLOYEES TABLE")
print("================================================")

df = pd.read_sql("SELECT * FROM employees", engine)

print(df)


# ============================================================
# STEP 6 — READ SPECIFIC COLUMNS
# ============================================================

print("\n================================================")
print("READING SPECIFIC COLUMNS")
print("================================================")

query = """
SELECT name, salary
FROM employees
"""

df_columns = pd.read_sql(query, engine)

print(df_columns)


# ============================================================
# STEP 7 — FILTER DATA
# ============================================================

print("\n================================================")
print("EMPLOYEES WITH SALARY > 70000")
print("================================================")

query = """
SELECT *
FROM employees
WHERE salary > 70000
"""

df_filtered = pd.read_sql(query, engine)

print(df_filtered)


# ============================================================
# STEP 8 — READ TABLE DIRECTLY
# ============================================================

print("\n================================================")
print("READ TABLE DIRECTLY USING read_sql_table()")
print("================================================")

df_table = pd.read_sql_table("employees", engine)

print(df_table)


# ============================================================
# STEP 9 — SHOW TABLE SCHEMA
# ============================================================

print("\n================================================")
print("TABLE SCHEMA")
print("================================================")

columns = inspector.get_columns("employees")

for column in columns:
    print(f"{column['name']} --> {column['type']}")


# ============================================================
# STEP 10 — BASIC DATA ANALYSIS USING PANDAS
# ============================================================

print("\n================================================")
print("BASIC DATA ANALYSIS")
print("================================================")

print("\nAverage Salary:")
print(df["salary"].mean())

print("\nMaximum Salary:")
print(df["salary"].max())

print("\nMinimum Salary:")
print(df["salary"].min())

print("\nDepartment-wise Average Salary:")
print(df.groupby("department")["salary"].mean())


# ============================================================
# STEP 11 — SORT DATA
# ============================================================

print("\n================================================")
print("SORT EMPLOYEES BY SALARY")
print("================================================")

sorted_df = df.sort_values(by="salary", ascending=False)

print(sorted_df)


# ============================================================
# STEP 12 — EXPORT TO CSV
# ============================================================

df.to_csv("employees_export.csv", index=False)

print("\nData exported to employees_export.csv")


# ============================================================
# STEP 13 — SQLALCHEMY CORE QUERY
# ============================================================

print("\n================================================")
print("SQLALCHEMY CORE QUERY")
print("================================================")

with engine.connect() as connection:

    result = connection.execute(
        text("""
        SELECT name, department
        FROM employees
        WHERE department = 'IT'
        """)
    )

    for row in result:
        print(row)


# ============================================================
# PROGRAM COMPLETED
# ============================================================

print("\nProgram completed successfully!")
```
